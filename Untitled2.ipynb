```python
import os
import requests
from tqdm import tqdm
import concurrent.futures

def download_molecule(zinc_id, zinc_version, zinc_file_type, output_dir):
    url = f"https://zinc{zinc_version}.docking.org/substances/{zinc_id}.{zinc_file_type}"
    file_path = os.path.join(output_dir, f"{zinc_id}.{zinc_file_type}")
    
    if os.path.exists(file_path):
        print(f"Skipping {zinc_id}.{zinc_file_type} (already downloaded)")
        return
    
    try:
        response = requests.get(url, timeout=10)
        if response.status_code == 200:
            with open(file_path, "wb") as f:
                f.write(response.content)
        else:
            print(f"Failed to download {zinc_id}.{zinc_file_type}")
    except requests.exceptions.RequestException as e:
        print(f"Error downloading {zinc_id}.{zinc_file_type}: {e}")

def get_all_zinc_ids():
    url = "https://zinc15.docking.org/substances.txt"  # This may need adjusting based on actual ZINC15 sources
    response = requests.get(url)
    if response.status_code == 200:
        return response.text.splitlines()
    else:
        print("Failed to fetch ZINC15 dataset IDs.")
        return []

def main():
    zinc_version = "15"  # Fixed to ZINC15
    file_types = ["sdf", "smi", "csv", "xml", "json"]  # Download all formats
    zinc_id_list = get_all_zinc_ids()
    
    if not zinc_id_list:
        print("No valid ZINC15 IDs retrieved.")
        return
    
    print(f"Downloading {len(zinc_id_list)} molecules in all formats...")
    
    for file_type in file_types:
        output_dir = f"zinc15_dataset/{file_type}"  # Organize by format
        os.makedirs(output_dir, exist_ok=True)
        
        with concurrent.futures.ThreadPoolExecutor(max_workers=os.cpu_count() * 2) as executor:
            futures = [executor.submit(download_molecule, zinc_id, zinc_version, file_type, output_dir) for zinc_id in zinc_id_list]
            for future in tqdm(concurrent.futures.as_completed(futures), total=len(futures), unit="molecules"):
                try:
                    future.result()
                except Exception as e:
                    print(f"Error: {e}")

if __name__ == '__main__':
    main()
```
