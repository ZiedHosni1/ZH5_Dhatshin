```python
import requests
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
from rdkit import Chem
from rdkit.Chem import AllChem
from torch.utils.data import Dataset, DataLoader

# -------------------------------
# 1️⃣ Fetch Molecules from ZINC15 API
# -------------------------------
def fetch_zinc15_bp_molecules():
    """
    Fetches molecules containing Boron (B) and Phosphorus (P) from ZINC15 API.
    """
    url = "https://zinc15.docking.org/substances/subsets/all/filters/element/B,P.csv"
    response = requests.get(url)

    if response.status_code == 200:
        data = response.text.split("\n")
        smiles_list = [line.split(",")[0] for line in data if line]  # Extract SMILES
        return smiles_list
    else:
        print(f"Failed to fetch data: {response.status_code}")
        return []

# Fetch molecules
bp_smiles = fetch_zinc15_bp_molecules()

# Convert to DataFrame and Save
bp_smiles_df = pd.DataFrame(bp_smiles, columns=["B/P Containing SMILES"])
bp_smiles_df.to_csv("filtered_bp_molecules.csv", index=False)
print("✅ Filtered Boron/Phosphorus molecules saved to 'filtered_bp_molecules.csv'")
print(bp_smiles_df.head())  # Preview first few entries
```

```python
# -------------------------------
# 2️⃣ Convert SMILES to Morgan Fingerprints (for WGAN training)
# -------------------------------
class ZINC15BPDataset(Dataset):
    def __init__(self, smiles_list):
        self.smiles = smiles_list
        self.fingerprints = [self.smiles_to_fp(smile) for smile in self.smiles]
    
    def smiles_to_fp(self, smiles):
        mol = Chem.MolFromSmiles(smiles)
        if mol:
            return np.array(AllChem.GetMorganFingerprintAsBitVect(mol, 2, nBits=1024))
        else:
            return np.zeros(1024)
    
    def __len__(self):
        return len(self.fingerprints)
    
    def __getitem__(self, idx):
        return torch.tensor(self.fingerprints[idx], dtype=torch.float32)

# Convert fetched molecules into dataset
zinc15_bp_dataset = ZINC15BPDataset(bp_smiles)

# Create DataLoader
data_loader = DataLoader(zinc15_bp_dataset, batch_size=64, shuffle=True)
```
