```python
from rdkit import Chem
import pandas as pd

# Expand dataset using key substructures (we'll vary R-groups and connectivity)
# Generate broader FLP motifs programmatically

# Common Lewis acid fragments
boron_cores = [
    "B(C)(C)C",  # Trialkylborane
    "B(O)(O)C",  # Boronic acid
    "B(C1=CC=CC=C1)(C)(C)",  # Arylborane
    "B(OC)(OC)OC",  # Borate ester
    "B(F)(F)F",  # BF3 analogue
]

# Common Lewis base fragments
base_cores = [
    "N(C)(C)C",  # Tertiary amine
    "NC(C)C",    # Secondary amine
    "N",         # Primary amine
    "P(C)(C)C",  # Trimethylphosphine
    "O=P(C)(C)C",  # Phosphine oxide
    "O",         # Hydroxy group
    "S",         # Thiol group
]

# Generate dot-separated FLP SMILES (acid.base)
flp_combinations = [f"{acid}.{base}" for acid in boron_cores for base in base_cores]

# Add silicon and tin acids
extra_acids = ["[Si](C)(C)(C)", "[Sn](C)(C)C"]
flp_combinations += [f"{acid}.{base}" for acid in extra_acids for base in base_cores]

# Add more diversity with aryl bases
aryl_bases = [
    "NC1=CC=CC=C1",     # Aniline
    "O=P(O)(O)N(C)C",   # Aminophosphate
    "C1=CC=CC=C1N",     # Phenylamine
]

flp_combinations += [f"{acid}.{base}" for acid in boron_cores + extra_acids for base in aryl_bases]

# Convert to DataFrame and deduplicate
df_flp = pd.DataFrame(sorted(set(flp_combinations)), columns=["smiles"])

# Save expanded dataset
expanded_file_path = "output/expanded_flp_smiles.txt"
df_flp.to_csv(expanded_file_path, index=False, header=False)

expanded_file_path
```
