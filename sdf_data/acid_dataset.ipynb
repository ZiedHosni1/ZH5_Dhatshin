```python
import pandas as pd
from rdkit import Chem
from rdkit.Chem import Descriptors

# Assume `filtered` is your list of valid SMILES
# Example: filtered = ["[B](c1ccccc1)(c1ccc(F)cc1)(c1ccc(Cl)cc1)", ...]

data = []
for smi in filtered:
    mol = Chem.MolFromSmiles(smi)
    if mol:
        mw = Descriptors.MolWt(mol)
        name = Chem.MolToSmiles(mol)  # Canonical SMILES
        data.append({
            "Name": name,
            "smiles": smi,
            "MolecularWeight": round(mw, 2)
        })

# Convert to DataFrame
df = pd.DataFrame(data)

# Save to CSV
df.to_csv("lewis_acid_candidates.csv", index=False)

print(f"Saved {len(df)} entries to lewis_acid_candidates.csv")
```
