```python
from rdkit import Chem
from rdkit.Chem import Descriptors
import pandas as pd

def read_sdf_to_dataframe(sdf_file):
    molecules = []
    
    # Read SDF file with strict=False to prevent crashes
    suppl = Chem.SDMolSupplier(sdf_file, sanitize=False, removeHs=False)

    for idx, mol in enumerate(suppl):
        if mol is None:
            print(f"Skipping molecule at index {idx} due to structure issues.")
            continue  # Skip invalid molecules
        
        try:
            mol_data = {}
            
            # Attempt to sanitize the molecule
            Chem.SanitizeMol(mol, sanitizeOps=Chem.SanitizeFlags.SANITIZE_ALL)
            
            # Extract molecular properties
            mol_data["SMILES"] = Chem.MolToSmiles(mol)  # Convert to SMILES
            mol_data["Molecular Weight"] = Descriptors.ExactMolWt(mol)  # Molecular weight
            
            # Extract all available properties in the SDF file
            for prop in mol.GetPropNames():
                mol_data[prop] = mol.GetProp(prop)
            
            molecules.append(mol_data)

        except Exception as e:
            print(f"Skipping molecule at index {idx} due to error: {e}")

    # Convert the list to a Pandas DataFrame
    df = pd.DataFrame(molecules)
    return df

# Example usage
sdf_file_1 = "FLP_datasetp1.sd"
sdf_file_2 = "FLP_datasetp2.sd"
df1 = read_sdf_to_dataframe(sdf_file_1)
df2 = read_sdf_to_dataframe(sdf_file_2)
```

```python
df1['count'] = df1.groupby(['SMILES','Molecular Weight']).cumcount()
df2['count'] = df2.groupby(['SMILES','Molecular Weight']).cumcount()
```

```python
combined_df = pd.concat([df1, df2], ignore_index=False)
```

```python
combined_df = combined_df.drop_duplicates()
```

```python
combined_df = combined_df.drop(['count'], axis=1)
```
