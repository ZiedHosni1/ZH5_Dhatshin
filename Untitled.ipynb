```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
import rdkit
from rdkit import Chem
from rdkit.Chem import Descriptors, rdMolDescriptors
from torch.utils.data import DataLoader, Dataset
from rdkit.Chem import AllChem
from ccdc.io import CrystalReader
from surface_charge_calculator import SurfaceChargeController
from hydrogen_bond_propensity_report import PropensityCalc
from show_semiconductor_properties import write_descs_report
```

```python
# --------------------
# 1️⃣ Phase 1: Pre-Training on ZINC15 (General Molecular Learning)
# --------------------
class ZINC15Dataset(Dataset):
    def __init__(self, smiles_list):
        self.smiles = smiles_list
        self.fingerprints = [self.smiles_to_fp(smile) for smile in self.smiles]
    
    def smiles_to_fp(self, smiles):
        mol = Chem.MolFromSmiles(smiles)
        if mol:
            return np.array(AllChem.GetMorganFingerprintAsBitVect(mol, 2, nBits=1024))
        else:
            return np.zeros(1024)
    
    def __len__(self):
        return len(self.fingerprints)
    
    def __getitem__(self, idx):
        return torch.tensor(self.fingerprints[idx], dtype=torch.float32)
```

```python
# Load ZINC15 dataset
zinc15_smiles = ["CCOC(=O)C1=CC=CC=C1", "CCC(=O)OCC", "CCCCCCCC"]  # Placeholder
zinc15_dataset = ZINC15Dataset(zinc15_smiles)
zinc15_loader = DataLoader(zinc15_dataset, batch_size=64, shuffle=True)
```

```python
# --------------------
# 2️⃣ Phase 2: Fine-Tuning on CSD (FLP-Specific Learning)
# --------------------
def extract_flp_smiles():
    """ Extracts FLP molecules from CSD using their refcodes """
    csd_reader = CrystalReader('CSD')
    flp_refcodes = ["FLP1", "FLP2", "FLP3"]  # Replace with real FLP refcodes
    smiles_list = []
    
    for refcode in flp_refcodes:
        try:
            mol = csd_reader.molecule(refcode)
            smiles_list.append(mol.smiles)
        except:
            print(f"Could not retrieve {refcode}")

    return smiles_list

flp_smiles = extract_flp_smiles()
csd_dataset = ZINC15Dataset(flp_smiles)
csd_loader = DataLoader(csd_dataset, batch_size=64, shuffle=True)
```

```python
# --------------------
# 3️⃣ GAN Training
# --------------------
class Generator(nn.Module):
    def __init__(self, latent_dim, output_dim):
        super(Generator, self).__init__()
        self.model = nn.Sequential(
            nn.Linear(latent_dim, 512),
            nn.ReLU(),
            nn.Linear(512, 1024),
            nn.ReLU(),
            nn.Linear(1024, output_dim),
            nn.Tanh()
        )
    def forward(self, z):
        return self.model(z)

class Discriminator(nn.Module):
    def __init__(self, input_dim):
        super(Discriminator, self).__init__()
        self.model = nn.Sequential(
            nn.Linear(input_dim, 1024),
            nn.ReLU(),
            nn.Linear(1024, 512),
            nn.ReLU(),
            nn.Linear(512, 1)
        )
    def forward(self, x):
        return self.model(x)

generator = Generator(latent_dim=128, output_dim=1024)
discriminator = Discriminator(input_dim=1024)
```
